# Cálculo de área

Notebook para praticar funções em Python calculando a área de algumas formas geométricas: retângulo, quadrado, círculo e triângulo.

## Retângulo

Área = comprimento × largura

```python
def calculate_area_rectangle(length, width):
    area = length * width
    return area

length = 5
width = 5
rectangle_area = calculate_area_rectangle(length, width)
print(f"A área do retângulo é: {rectangle_area}")
```

## Quadrado

Área = lado²

```python
def calculate_area_square(side):
    area = side ** 2
    return area

side = 4
square_area = calculate_area_square(side)
print(f"A área do quadrado é: {square_area}")
```

## Círculo

Área = π × raio²

```python
import math

def calculate_area_circle(radius):
    area = math.pi * radius ** 2
    return area

radius = 3
circle_area = calculate_area_circle(radius)
print(f"A área do círculo é: {circle_area:.2f}")
```

## Triângulo

Área = (base × altura) / 2

```python
def calculate_area_triangle(base, height):
    area = (base * height) / 2
    return area

base = 6
height = 4
triangle_area = calculate_area_triangle(base, height)
print(f"A área do triângulo é: {triangle_area}")
```

## Resumo

Reunindo todos os cálculos em um só lugar.

```python
print(f"Retângulo: {rectangle_area}")
print(f"Quadrado: {square_area}")
print(f"Círculo: {circle_area:.2f}")
print(f"Triângulo: {triangle_area}")
```
