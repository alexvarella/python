```python
# Trocando a classe para somar string com integer - não consegue somar nome com número, para isso na variável idade deve-se mudar para a class int (integer)
nome = input('Digite o seu nome: ')
idade = int(input('Digite a sua idade: '))

# f-string
print(f'Olá {nome}, você tem {idade} anos e daqui a 5 anos você terá {idade + 5}.')
# print(type(nome))
# print(type(idade))
```

```python
# cálculos matemáticos

num1 = 10
num2 = 3
print(num1 + num2)
print(num1 - num2)
print(num1 * num2)
print(num1 / num2) # divisão
print(num1 // num2) # divisão inteira
print(num1 % num2) # resto da divisão
print(num1 ** num2) # potência
```

```python
# cálculo de desconto

preco = int(50) # valor em R$
desconto = int(10) # desconto em %
novo_preco = preco - preco * desconto / 100
print(f'Calça, {preco}. Pagamento a vista {novo_preco} ')

```

```python
preco = int(input('Digite o valor do produto: '))
desconto = int(input('Digite o desconto a ser aplicado '))
novo_preco = preco - preco * desconto / 100
print(f'O valor com desconte é de R$ {novo_preco}')
```

```python
# Calcular quantos dias um produto X duraria se a pessoa usar X porções por dia:

produto = int(input('Quantas porções tem o produto? '))
dias = int(input('Quantas porções você vai usar por dia? '))
duracao = produto / dias
print(f'O produto irá durar {duracao:.0f} dias!')
```

```python
# Operadores relacionais: É usado para comparar valores e relacionar se é TRUE ou FALSE

num1 = 10
num2 = 5
num3 = 10
print(num1 == num3) # == Perguntando se esses valores são iguais
print(num1 != num2) # != Perguntando se os valores são diferentes
print(num1 > num2) # > Maior que
print(num1 < num2) # < Menor que
print(num1 >= num2) # >= Maior ou igual
print(num1 <= num2) # <= Menor ou igual



```
